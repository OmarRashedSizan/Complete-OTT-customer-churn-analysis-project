# OTT Customer Churn — Business Insights

## Executive Summary

This analysis evaluates customer churn patterns across subscription, customer, and support-related factors.

The dataset contains **600 customers**, with an overall churn rate of **54.83%**.

The analysis identified several high-risk customer segments, particularly customers on **Basic plans**, **Monthly contracts**, and certain **plan-contract combinations**.

Key findings indicate that:

- Basic-plan customers have the highest observed churn rate (**81.94%**).
- Monthly-contract customers have a substantially higher churn rate (**70.27%**) than Annual-contract customers (**39.80%**).
- Basic + Monthly customers show an **82.86%** churn rate.
- Premium + Monthly customers show a **100%** churn rate within the observed 62-customer segment.
- Referral-acquired customers have a **61.18%** churn rate.
- Active customers have substantially higher median CLTV (**935**) than churned customers (**555**).

These findings suggest that retention efforts should focus on identifying high-risk customer segments and prioritizing targeted interventions rather than applying the same retention strategy to all customers.

> **Note:** These findings represent observed associations in the dataset and should not be interpreted as causal relationships.

## 1. Overall Churn Situation

The dataset contains 600 customers:

- **329 Churned**
- **271 Active**
- **Overall Churn Rate: 54.83%**

A churn rate above half of the customer base indicates a significant customer retention challenge.

### Business Implication

The business should prioritize identifying high-risk customer segments and developing targeted retention strategies rather than treating all customers equally.

## 2. Highest-Risk Customer Segments

The analysis identified several customer segments with substantially higher observed churn rates.

| Segment | Churn Rate |
|---|---:|
| Basic Plan | **81.94%** |
| Monthly Contract | **70.27%** |
| Referral Acquisition | **61.18%** |
| Basic + Monthly | **82.86%** |
| Premium + Monthly | **100.00%** |

### Business Interpretation

The strongest observed risk appears in specific combinations of plan and contract type.

In particular, **Basic + Monthly** customers represent a large high-risk segment with 82.86% observed churn.

The **Premium + Monthly** segment also shows 100% observed churn among 62 customers and therefore deserves investigation.

These segments should be prioritized for retention analysis and intervention.

## 3. Key Business Drivers

The analysis indicates that churn is most strongly associated with specific subscription characteristics rather than a single customer attribute.

### 1. Contract Commitment

Monthly-contract customers show substantially higher churn than Annual-contract customers.

This suggests that customers with lower contractual commitment may be more vulnerable to churn.

### 2. Plan Type

Basic-plan customers have the highest observed churn rate among the three plans.

This indicates a need to investigate the Basic plan's pricing, features, content value, and customer expectations.

### 3. Plan × Contract Combination

The combination of plan type and contract type reveals particularly high-risk segments.

Basic + Monthly customers have an 82.86% churn rate, while Premium + Monthly customers show 100% observed churn among 62 customers.

### 4. Acquisition Type

Referral-acquired customers show a higher churn rate than Organic and Paid customers.

This suggests that the business should investigate whether referred customers have different expectations, onboarding experiences, or retention behavior.

### 5. Customer Lifetime Value

Active customers have substantially higher CLTV than churned customers.

This suggests that CLTV can be useful as a prioritization metric when designing retention strategies.

### Overall Interpretation

The findings suggest that churn risk is better understood through a combination of customer characteristics rather than a single variable.

The highest priority should therefore be given to identifying high-risk subscription segments and combining churn risk with customer value when designing retention actions.

## 4. Business Recommendations

Based on the observed churn patterns, the following business actions are recommended.

### 1. Prioritize Monthly Customers for Retention

Monthly-contract customers have a 70.27% observed churn rate.

The business should introduce targeted retention campaigns for high-risk Monthly customers, especially before renewal or cancellation periods.

### 2. Investigate the Basic Plan

The Basic plan has an 81.94% churn rate.

The business should review:

- Pricing and perceived value
- Available content and features
- Customer expectations
- Upgrade and downgrade behavior

The objective should be to understand why Basic customers are leaving at a higher rate.

### 3. Target High-Risk Plan-Contract Segments

Basic + Monthly customers have an 82.86% churn rate.

Premium + Monthly customers show 100% observed churn among 62 customers.

These segments should receive priority in retention monitoring and further investigation.

### 4. Review Referral Customer Retention

Referral customers have a 61.18% observed churn rate.

The business should investigate the onboarding experience, expectations, and long-term behavior of referred customers.

### 5. Combine Churn Risk with Customer Value

Active customers have substantially higher CLTV than churned customers.

Retention efforts should therefore consider both:

- Probability of churn
- Customer lifetime value

High-value customers with elevated churn risk should receive higher retention priority.

### Expected Business Impact

Implementing targeted retention strategies can help the business:

- Identify high-risk customer segments earlier
- Focus retention resources where they are most valuable
- Reduce unnecessary broad-based retention campaigns
- Protect high-value customers
- Improve customer retention and long-term revenue potential

## 5. Analysis Limitations

The findings should be interpreted within the limitations of the available dataset.

- The analysis is based on 600 customers and represents the available dataset rather than the entire customer population.
- The identified relationships are observational and do not establish causation.
- Support-related analysis was limited to customers with available support records.
- Some data quality issues and anomalies were identified during the cleaning process, particularly in monthly charges.
- Some customer segments have relatively small sample sizes and should therefore be interpreted cautiously.
- Additional behavioral, financial, and product-level data could improve future churn modeling.

## 6. Final Conclusion

The analysis of 600 customers identified a significant customer retention challenge, with an overall churn rate of **54.83%**.

The strongest observed churn patterns were associated with **Basic plans, Monthly contracts, specific plan-contract combinations, and Referral acquisition**. In particular, Basic + Monthly customers showed an **82.86% churn rate**, while the Premium + Monthly segment showed **100% observed churn among 62 customers**.

The analysis also found that active customers have substantially higher CLTV than churned customers, suggesting that customer value can be an important consideration when prioritizing retention efforts.

On the other hand, **support escalation count and CSAT did not show strong or consistent differences in churn** within the available data. Therefore, these factors should not be used independently as primary churn indicators.

### Final Business Takeaway

The business should move from a broad, one-size-fits-all retention strategy toward a **targeted customer retention approach**.

Priority should be given to:

1. High-risk Monthly-contract customers
2. Basic-plan customers
3. High-risk plan-contract combinations
4. Referral-acquired customers
5. High-value customers with elevated churn risk

The next step would be to develop a **customer-level churn prediction and risk segmentation system** that combines subscription characteristics, customer value, and behavioral signals to identify customers who are most likely to churn.

> **Overall conclusion:** The analysis provides a clear foundation for targeted retention strategies, while further predictive modeling and additional behavioral data could improve the accuracy of churn-risk identification.
